```python
# Import libraries
import fitz # PyMuPDF library for reading PDF files and extracting images 
from PIL import Image  # Pillow library for image processing
import os
```

Defining a function to pdf to jpg conversion

```python
def pdf_to_jpg(directory):
  # Go through the directory and find all files with pdf ending
  for filename in os.listdir(directory):
    if filename.endswith('.pdf'): # Check if the file is a PDF file
      pdf_path = os.path.join(directory, filename) # Create the path to the PDF file
      pdf_document = fitz.open(pdf_path) # Open the PDF
      #pdf_document = fitz.open(pdf_path) # Open the PDF
      for page_number in range(len(pdf_document)): # Go through all pages in the PDF
        page = pdf_document.load_page(page_number) # Load the page
        pix = page.get_pixmap() # Get the image of the page

        # Save image directly in the given directory
        image_path = os.path.join(directory, f"{os.path.splitext(filename)[0]}_page_{page_number +1}.jpg") # Create the path to the image file
        img = Image.frombytes("RGB", [pix.width, pix.height], pix.samples) # Create the image from the bytes of the image
        img.save(image_path) # Save the image

  print("All PDF files have been converted")
```

```python
# Apply the function
pdf_to_jpg(directory)
```
